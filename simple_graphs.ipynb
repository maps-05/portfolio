```python
# Line Graph Customization

a = [1.0, 1.25, 1.5, 1.75, 2.0, 2.25, 2.5, 2.75, 3.0, 4.0, 5.0]
b = [96, 93, 90, 87, 84, 81, 78, 75, 70, 65, 0]

# line style and color
plt.plot(a, b, marker = 'o', linestyle='-.', color='red')
plt.title('Grade Conversion')
plt.xlabel('Grade Point')
plt.ylabel('Percentage (%)')

plt.grid(True)
plt.show()
```

```python
# Customizing Bar graphs

x = [1.0, 1.25, 1.5, 1.75, 2.0, 2.25, 2.5, 2.75, 3.0, 4.0, 5.0]
y = [96, 93, 90, 87, 84, 81, 78, 75, 70, 65, 0]

# custom colors for each bar
colors = ['green', 'lime', 'yellowgreen', 'gold', 'orange', 'salmon',
          'red', 'brown', 'maroon', 'purple', 'gray']

plt.bar(x, y, color=colors, width=0.15)
plt.title('Grade Conversion')
# labels
plt.xlabel('Grade Point')
plt.ylabel('Percentage (%)')
# grid lines
plt.grid(True, which='both', linestyle='--', linewidth=0.5, color='gray')
plt.show()
```

```python
# Adding Customizations

x = [1.0, 1.25, 1.5, 1.75, 2.0, 2.25, 2.5, 2.75, 3.0, 4.0, 5.0]
y = [96, 93, 90, 87, 84, 81, 78, 75, 70, 65, 0]

colors = ['green', 'lime', 'yellowgreen', 'gold', 'orange', 'salmon',
          'red', 'darkred', 'brown', 'gray', 'black']

fig, (ax1, ax2) = plt.subplots(2, 1, figsize=(10, 10))

ax1.plot(x, y, marker='o', linestyle='-', color='green', label='Percentage (%)')
ax1.set_title('Line Graph of Grade Conversion')
ax1.set_xlabel('Grade Point')
ax1.set_ylabel('Percentage (%)')
# grid line customization
ax1.grid(True, linestyle='--', linewidth=0.5, color='blue', alpha=0.5)
# legend
ax1.legend()

# annotate the highest data point
max_y = max(y)
max_x = x[y.index(max_y)]
ax1.text(max_x, max_y, f'{max_y}', fontsize=10, ha='center', va='bottom')

bars = ax2.bar(x, y, color=colors, width=0.15)
ax2.set_title('Bar Graph of Grade Conversion')
ax2.set_xlabel('Grade Point')
ax2.set_ylabel('Percentage (%)')
ax2.set_xticks(np.arange(1.0, 5.1, 0.25))
# gridline customization
ax2.grid(True, which='both', linestyle='--', linewidth=0.5, color='blue', alpha=0.5)
# legend
ax2.legend(['Percentage (%)'])

# annotate the highest data point
max_y_bar = max(y)
max_x_bar = x[y.index(max_y_bar)]
ax2.text(max_x_bar, max_y_bar, f'{max_y_bar}', fontsize=10, ha='center', va='bottom')

plt.tight_layout()
plt.show()
```

DISCUSSION QUESTIONS

1. When would you choose a bar graph over a line graph, and why?

I would choose a bar graph over a line graph when I want to compare discrete categories or groups because bar graphs clearly show DIFFERENCES IN SIZE OR FREQUENCY among distinct items. It is also ideal for problems involving data sets that do not have a continuous relationship with each other. By illustrating differences in size or frequency among items, it makes the graph easier to visualize and interpret categorical data.

2. What are the some advantages of using subplots in a single figure?

Using subplots in a single figure allows for the comparison of multiple datasets side by side, making it easier to identify relationship or connection among them. It helps organize visual information more efficiently, providing a clear and concise presentation without overcrowding a single plot. Furthermore, subplots can enhance the overall readability of complex data visualizations.

3. How does customizing graphs improve the clarity of data visualization?

For me customizing graphs adds data visualization clarity by making the information more suitable for the normal audience. By adjusting elements like colors, labels, and scales, it becomes easier to highlight key insights and ensure that important trends or key points are easily seen and understood. Another thing is that personalized graphs can reduce confusion and prevent misinterpretation by presenting the data in a clearer way.
