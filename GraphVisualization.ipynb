```python
import matplotlib.pyplot as plt
import numpy as np

# Data
factors = ["Slow Order Taking", "Order Taking Errors", "Understaff", "Machine Malfunction", "Delivery Delays"]
frequency = [48, 14, 9, 6, 3]
percentage = [60, 17.5, 11.25, 7.5, 3.75]
cumulative_percentage = [60, 77.5, 88.75, 96.25, 100]
colors = ["red", "orange", "yellow", "green", "brown"]

fig, ax1 = plt.subplots(figsize=(10, 6), dpi=300)

bars = ax1.bar(factors, frequency, color=colors, label='Frequency')
ax1.set_ylabel('Frequency', color='blue', fontsize=12, fontweight='bold')
ax1.tick_params(axis='y', labelcolor='blue')
ax1.set_ylim(0, 80)
ax1.set_yticks(np.arange(0, 81, 20))

for bar, freq in zip(bars, frequency):
    ax1.text(bar.get_x() + bar.get_width() / 2, -5, f'{freq}%', ha='center', va='top', fontsize=10)

ax2 = ax1.twinx()
ax2.set_ylim(0, 100)
ax2.set_yticks(np.arange(0, 101, 10))
ax2.plot(factors, cumulative_percentage, color='red', marker='o', linestyle='-', linewidth=2, label='Cumulative Percentage')
ax2.set_ylabel('Percentage (%)', color='red', fontsize=12, fontweight='bold')
ax2.tick_params(axis='y', labelcolor='red')

for i, perc in enumerate(cumulative_percentage):
    ax2.text(i, perc + 1, f'{perc}%', ha='center', va='bottom', color='red', fontsize=9)

ax1.grid(axis='y', linestyle='-', alpha=0.5)
ax1.set_title('Causes of Late Order Completion', fontsize=20, fontweight='bold')
ax1.legend(loc='upper left', fontsize=8)
ax2.legend(loc='upper right', fontsize=8)

fig.tight_layout()
plt.show()
```
