```python
# Load Dataset

df = pd.read_csv("https://raw.githubusercontent.com/renatomaaliw3/public_files/refs/heads/master/Data%20Sets/cxp_crypto.csv", index_col='Date', parse_dates=True)
df
```

QUESTION NO. 3

```python
df_weekly = df["Close"].resample("W").mean()

value_3 = df_weekly.loc["2016-05-01"]
value_3.round(2)
```

QUESTION NO. 4

```python
df_quarterly = df["Close"].resample("QE").median()

value_4 = df_quarterly.loc["2017-12-31"]
value_4.round(2)
```

QUESTION NO. 5

```python
df_semi_monthly = df["Volume"].resample("SME").mean()

last_five = df_semi_monthly.tail(5)
median_value = last_five.median()
median_value = round(median_value, 2)
median_value
```

QUESTION NO. 6

```python
df['Close'].resample('QE').max().plot(kind='bar', color='red', rot=90, figsize=(15,6), title="Quarterly Max Closing Price");
```

QUESTION NO. 7

```python
df_shifted = df.shift(periods=17)

shift_for = df_shifted.loc['2015-02-26', 'Close']
shift_for.round(2)
```

QUESTION NO. 8

```python
df_shifted = df.shift(periods=-28)

shift_back = df_shifted.loc['2018-08-31', 'Close']
shift_back.round(2)
```

QUESTION NO. 9

```python
df['Close_14_days'] = df['Close'].rolling(window=14).mean()

close_price = df.loc['2015-04-09', 'Close_14_days']
close_price.round(2)
```

QUESTION NO. 10

```python
df['Close: 7 days'] = df['Close'].rolling(window=7).mean()
df['Close: 14 days'] = df['Close'].rolling(window=14).mean()
df['Close: 21 days'] = df['Close'].rolling(window=21).mean()
df['Close: 30 days'] = df['Close'].rolling(window=30).mean()

df[['Close', 'Close: 7 days', 'Close: 14 days', 'Close: 21 days', 'Close: 30 days']].plot(figsize=(15,10), title="CXP Closing Price");
```

QUESTION NO. 11

**A smaller rolling window makes the moving average more responsive to short-term price changes but also more volatile. On the other hand, a larger window smooths out fluctuations, making long-term trends clearer but slower to react to recent movements. Choosing the right window size depends on whether the focus is on short-term market changes or overall trend direction.**
