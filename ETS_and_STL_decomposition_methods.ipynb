QUESTION 1

```python
from statsmodels.tsa.seasonal import seasonal_decompose

decomposition6 = seasonal_decompose(df_class_06['Value'], model='additive')

plt.figure(figsize=(30, 5))
decomposition6.seasonal.plot()
plt.title('Seasonal Component of the Classified TS 6')
plt.xlabel('Date')
plt.ylabel('Seasonal Value')
plt.grid(True)
plt.show()
```

QUESTION 2

```python
plt.figure(figsize=(30, 5))
decomposition6.trend.plot()
plt.title("Trend Component of the Classified TS 6")
plt.xlabel("Date")
plt.ylabel("Trend Value")
plt.grid(True)
plt.show()
```

QUESTION 6

```python
from statsmodels.tsa.seasonal import STL

stl8 = STL(df_class_08["Value"], robust = True)
decomp_stl_8 = stl8.fit()

plt.figure(figsize=(30, 5))
decomp_stl_8.resid.plot()
plt.title('Residual Component of the Classified TS 8')
plt.xlabel('Date')
plt.ylabel('Residual Value')
plt.grid(True)
plt.show()
```

QUESTION 7

```python
plt.figure(figsize=(30, 5))
decomp_stl_8.trend.plot()
plt.title('Trend Component of the Classified TS 8')
plt.xlabel('Date')
plt.ylabel('Trend Value')
plt.grid(True)
plt.show()
```

QUESTION 8

```python
zoomed_data = decomp_stl_8.seasonal['1980-03-01':'1984-12-31']

plt.figure(figsize=(30, 5))
zoomed_data.plot()
plt.title('Zoomed-In Seasonal Component (2010-2013) of the Classified TS 8')
plt.xlabel('Date')
plt.ylabel('Seasonal Value')
plt.grid(True)
plt.show()
```

QUESTION 12

```python
ets_result6 = seasonal_decompose(df_class_06['Value'], model='additive')
ets_result7 = seasonal_decompose(df_class_07['Value'], model='additive')

stl_result6 = STL(df_class_06['Value'], period=4, robust=True).fit()
stl_result7 = STL(df_class_07['Value'], period=4, robust=True).fit()

fig, axes = plt.subplots(4, 4, figsize=(35, 20))

fig.suptitle("Comparative Graphs: ETS vs STL Decomposition", fontsize=15)

axes[0, 0].plot(df_class_06['Value'])
axes[0, 0].set_title("Original Series (TS6)")
axes[0, 0].grid(True)
axes[0, 1].plot(df_class_06['Value'])
axes[0, 1].set_title("Original Series (TS6)")
axes[0, 1].grid(True)
axes[0, 2].plot(df_class_07['Value'])
axes[0, 2].set_title("Original Series (TS7)")
axes[0, 2].grid(True)
axes[0, 3].plot(df_class_07['Value'])
axes[0, 3].set_title("Original Series (TS7)")
axes[0, 3].grid(True)

axes[1, 0].plot(ets_result6.trend)
axes[1, 0].set_title("ETS Trend (TS6)")
axes[1, 0].grid(True)
axes[1, 1].plot(stl_result6.trend)
axes[1, 1].set_title("STL Trend (TS6)")
axes[1, 1].grid(True)
axes[1, 2].plot(ets_result7.trend)
axes[1, 2].set_title("ETS Trend (TS7)")
axes[1, 2].grid(True)
axes[1, 3].plot(stl_result7.trend)
axes[1, 3].set_title("STL Trend (TS7)")
axes[1, 3].grid(True)

axes[2, 0].plot(ets_result6.seasonal)
axes[2, 0].set_title("ETS Seasonality (TS6)")
axes[2, 0].grid(True)
axes[2, 1].plot(stl_result6.seasonal)
axes[2, 1].set_title("STL Seasonality (TS6)")
axes[2, 1].grid(True)
axes[2, 2].plot(ets_result7.seasonal)
axes[2, 2].set_title("ETS Seasonality (TS7)")
axes[2, 2].grid(True)
axes[2, 3].plot(stl_result7.seasonal)
axes[2, 3].set_title("STL Seasonality (TS7)")
axes[2, 3].grid(True)

axes[3, 0].plot(ets_result6.resid)
axes[3, 0].set_title("ETS Residuals (TS6)")
axes[3, 0].grid(True)
axes[3, 1].plot(stl_result6.resid)
axes[3, 1].set_title("STL Residuals (TS6)")
axes[3, 1].grid(True)
axes[3, 2].plot(ets_result7.resid)
axes[3, 2].set_title("ETS Residuals (TS7)")
axes[3, 2].grid(True)
axes[3, 3].plot(stl_result7.resid)
axes[3, 3].set_title("STL Residuals (TS7)")
axes[3, 3].grid(True)

plt.tight_layout()
plt.show()
```
