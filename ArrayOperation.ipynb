2. BASIC ARRAY OPERATIONS

```python
array1 = np.array([[1, 3, 5], [7, 9, 11], [13, 15, 17]])
array2 = np.array([[2, 4, 6], [8, 10, 12], [14, 16, 18]])

# Element-wise Addition
addition = array1 + array2
print("Using element-wise Addition:")
print(addition)
```

```python
# Element-wise Subtraction
subtraction = array1 - array2
print("Using element-wise Subtraction:")
print(subtraction)
```

```python
# Element-wise Multiplication
multiplication = array1 * array2
print("Using element-wise Multiplication:")
print(multiplication)
```

```python
# Element-wise Division
division = array1 / array2
print("Using element-wise Division:")
print(division)
```

```python
# Extracting Specific Elements
spec_element = array1[1, 2]
print("Using specific element (array1[1, 2]):", spec_element)
```

```python
# Extracting a Subarray
subarray = array1[:2, :2]
print("Using subarray (first two rows and columns):")
print(subarray)
```

3. RESHAPING AND TRANSPOSING ARRAYS

```python
third_array = np.arange(1, 10)
print("Original Array:")
print(third_array)
```

```python
reshaped_array = third_array.reshape(3, 3)
print("Reshaped 3x3 Array:")
print(reshaped_array)
```

```python
transposed_array = reshaped_array.transpose()
print("Transposed Array:")
print(transposed_array)
```

4. ACTIVITY

```python
# Importing Numpy

import numpy as np
```

```python
# Creating a 3by3 Numpy array

my_array = np.arange(1, 10).reshape(3, 3)
print(my_array)
```

```python
# Finding the sum of the array

array_sum = np.sum(my_array)
print(f"The sum of this array is {array_sum}.")
```

```python
# Finding the mean of the array

array_mean = np.mean(my_array)
print(f"The mean of this array is {array_mean}.")
```

```python
# Finding the standard deviation of the array

array_std = np.std(my_array)
print(f"The standard deviation of this array is {array_std}.")
```

```python
# Extracting the subarray

subarray = [row[:2] for row in my_array[:2]]
print(subarray)
```
