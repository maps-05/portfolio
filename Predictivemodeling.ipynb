# Predictive Modeling: Regression and Classification Analysis

Midterm and Finals Exam

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error
```

```python
# Load dataset
df = pd.read_csv("https://raw.githubusercontent.com/renatomaaliw3/public_files/refs/heads/master/Data%20Sets/linear-m08.csv")
df
```

```python
print("Missing values:\n", df.isnull().sum())
```

```python
x = df['X']
y = df['Y']
```

```python
X_train, X_test, y_train, y_test = train_test_split(x, y, test_size=0.25, random_state=101)

# Reshape inputs to 2D arrays
X_train = np.array(X_train).reshape(-1, 1)
X_test = np.array(X_test).reshape(-1, 1)
model = LinearRegression()
model.fit(X_train, y_train)
```

```python
# Slope
slope = model.coef_[0]
print("Slope of the line:", slope)
```

```python
y_pred = model.predict(X_test)

# RMSE
rmse = np.sqrt(mean_squared_error(y_test, y_pred))
print("RMSE on test data:", rmse)
```

```python
# Predict Y for X = 38.85
x_input = np.array([[38.85]])
y_pred_input = model.predict(x_input)[0]
print("Predicted Y for X = 38.85:", y_pred_input)
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np

from sklearn.model_selection import train_test_split
from sklearn.svm import SVC
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import accuracy_score, classification_report
```

```python
# Load Dataset

df = pd.read_csv('https://raw.githubusercontent.com/renatomaaliw3/public_files/refs/heads/master/Data%20Sets/classification-f08.csv')
df
```

```python
df_base = df.iloc[:495]

# Calculate conditional means for Z = 0 and Z = 1
means_z0 = df_base[df_base['Z'] == 0].iloc[:, :-1].mean()
means_z1 = df_base[df_base['Z'] == 1].iloc[:, :-1].mean()

# Replace missing values in index 495 to 499 based on Z value
for idx in range(495, 500):
    z_value = df.loc[idx, 'Z']
    if z_value == 0:
        df.loc[idx, df.columns[:-1]] = df.loc[idx, df.columns[:-1]].fillna(means_z0)
    elif z_value == 1:
        df.loc[idx, df.columns[:-1]] = df.loc[idx, df.columns[:-1]].fillna(means_z1)

df
```

```python
from sklearn.model_selection import train_test_split, GridSearchCV
from sklearn.svm import SVC
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import accuracy_score, classification_report

X = df.iloc[:, :-1]
y = df['Z']

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.25, random_state=101)

# Scale the data
scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)

param_grid = {
    'C': [0.01, 0.05, 0.1, 0.5, 1, 1.5, 2, 2.5, 3, 3.5, 4, 4.5, 5],
    'kernel': ['linear', 'rbf', 'poly']
}

svc = SVC()
grid = GridSearchCV(svc, param_grid)
grid.fit(X_train_scaled, y_train)

best_model = grid.best_estimator_
y_pred = best_model.predict(X_test_scaled)
accuracy = accuracy_score(y_test, y_pred)
report = classification_report(y_test, y_pred)

# Predict for new data [30, 25, 35]
new_input = np.array([[30, 25, 35]])
new_input_scaled = scaler.transform(new_input)
prediction = best_model.predict(new_input_scaled)

# Print results
print("Optimized value of C:", grid.best_params_['C'])
print("Best kernel:", grid.best_params_['kernel'])
print("Accuracy on test set:", accuracy)
print("Classification Report:\n", report)
print("Prediction for [30, 25, 35]:", prediction[0])
```

```python
from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import accuracy_score, classification_report

dt = DecisionTreeClassifier(random_state=42)
dt.fit(X_train, y_train)

y_pred_dt = dt.predict(X_test)
accuracy_dt = accuracy_score(y_test, y_pred_dt)
report_dt = classification_report(y_test, y_pred_dt)

importances = dt.feature_importances_
feature_names = X.columns
if dt.tree_.feature[0] != -2:
    root_node = feature_names[dt.tree_.feature[0]]
else:
    root_node = "Cannot determine root node for this tree."

least_important = feature_names[np.argmin(importances)]

# Prediction for [35, 30, 25]
new_input_dt = np.array([[35, 30, 25]])
prediction_dt = dt.predict(new_input_dt)

print("Accuracy (Decision Tree):", accuracy_dt)
print("Classification Report (Decision Tree):\n", report_dt)
print("Root node feature:", root_node)
print("Least important feature:", least_important)
print("Prediction for new input [35, 30, 25]:", prediction_dt[0])
```

```python
from sklearn.ensemble import RandomForestClassifier
from sklearn.model_selection import GridSearchCV
from collections import Counter
import numpy as np


param_grid_rf = {
    'n_estimators': [11, 25, 65, 101, 151, 201, 301, 401],
    'max_features': [2, 3]
}

rf = RandomForestClassifier(random_state=42)
grid_rf = GridSearchCV(rf, param_grid_rf, cv=5, n_jobs=-1)
grid_rf.fit(X_train, y_train)

best_rf = grid_rf.best_estimator_
y_pred_rf = best_rf.predict(X_test)
accuracy_rf = accuracy_score(y_test, y_pred_rf)
report_rf = classification_report(y_test, y_pred_rf)

root_nodes = [tree.tree_.feature[0] for tree in best_rf.estimators_]
root_node_counts = Counter([X.columns[i] for i in root_nodes if i != -2])
most_common_root = root_node_counts.most_common(1)[0][0]

importances_rf = best_rf.feature_importances_
least_important_rf = X.columns[np.argmin(importances_rf)]

# Prediction for [35, 25, 30]
new_input_rf = np.array([[35, 25, 30]])
prediction_rf = best_rf.predict(new_input_rf)

print("Accuracy (Random Forest):", accuracy_rf)
print("Classification Report (Random Forest):\n", report_rf)
print("Most common root node feature:", most_common_root)
print("Least important feature:", least_important_rf)
print("Prediction for new input [35, 25, 30]:", prediction_rf[0])
```
